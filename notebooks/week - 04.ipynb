```python
# WEEK 1 — DATASET SETUP + EDA
import zipfile
import os
import matplotlib.pyplot as plt
import cv2
import random

# 1. UNZIP DATASET
# Путь к zip файлу
zip_path = r"C:\Users\Huawei\Downloads\archive (10).zip"

# Куда распаковать
extract_path = r"C:\Users\Huawei\Downloads\intel_dataset"

# Распаковка
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully!")

# 2. SET TRAIN PATH
train_path = os.path.join(extract_path, "seg_train", "seg_train")

# 3. CHECK CLASSE
classes = os.listdir(train_path)
print("Classes found:", classes)

# 4. VISUALIZE SAMPLE IMAGES
plt.figure(figsize=(12, 8))
for i, cls in enumerate(classes):
    class_path = os.path.join(train_path, cls)
    
    img_name = random.choice(os.listdir(class_path))
    img_path = os.path.join(class_path, img_name)

    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.subplot(2, 3, i + 1)
    plt.imshow(img)
    plt.title(cls)
    plt.axis("off")

plt.tight_layout()
plt.show()

# 5. DATA INSIGHT
# Количество изображений в каждом классе
for cls in classes:
    path = os.path.join(train_path, cls)
    print(cls, ":", len(os.listdir(path)), "images")

# COMMENT:
# This step:
# - loads dataset from ZIP file
# - extracts it
# - explores class structure
# - visualizes sample images
# - checks dataset balance 
```

```python
# WEEK 2-3 — CNN MODEL TRAINING

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import os 

# 1. DATASET PATH
extract_path = r"C:\Users\Huawei\Downloads\intel_dataset"

train_path = os.path.join(extract_path, "seg_train", "seg_train")
test_path = os.path.join(extract_path, "seg_test", "seg_test")

# 2. PARAMETERS
IMG_SIZE = 150
BATCH_SIZE = 32

# 3. DATA PREPROCESSING + AUGMENTATION
train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.1,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(rescale=1./255)

# Train set
train_generator = train_datagen.flow_from_directory(
    train_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training'
)

# Validation set
validation_generator = train_datagen.flow_from_directory(
    train_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation'
)

# Test set
test_generator = test_datagen.flow_from_directory(
    test_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# 4. CNN MODEL (FIXED VERSION)
model = models.Sequential()

# Input layer (fix warning)
model.add(layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3)))

# Conv Block 1
model.add(layers.Conv2D(32, (3,3), activation='relu'))
model.add(layers.MaxPooling2D(2,2))

# Conv Block 2
model.add(layers.Conv2D(64, (3,3), activation='relu'))
model.add(layers.MaxPooling2D(2,2))

# Conv Block 3
model.add(layers.Conv2D(128, (3,3), activation='relu'))
model.add(layers.MaxPooling2D(2,2))

# 5. CLASSIFIER
model.add(layers.Flatten())
model.add(layers.Dense(128, activation='relu'))
model.add(layers.Dropout(0.5))
model.add(layers.Dense(6, activation='softmax'))

# 6. COMPILE MODEL

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# 7. TRAIN MODEL

history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=10
)

# 8. EVALUATION ON TEST SET

test_loss, test_acc = model.evaluate(test_generator)
print("Test Accuracy:", test_acc)

# 9. ACCURACY GRAPH

plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend(["Train", "Validation"])
plt.show()

# 10. LOSS GRAPH
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend(["Train", "Validation"])
plt.show() 

# COMMENT:
# Week 2 includes:
# - preprocessing (normalization + split)
# - baseline CNN model
# - training process
# - evaluation on test set
# - accuracy/loss visualization 
```

```python
# WEEK 4 — FINAL MODEL EVALUATION + REPORT + EXPORT

import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import os

# 1. CLASS LABELS
class_labels = list(test_generator.class_indices.keys())

# 2. PREDICTIONS ON TEST SET
predictions = model.predict(test_generator)
predicted_classes = np.argmax(predictions, axis=1)
true_classes = test_generator.classes

# 3. CLASSIFICATION REPORT
print("\nCLASSIFICATION REPORT:\n")

report = classification_report(
    true_classes,
    predicted_classes,
    target_names=class_labels
)

print(report)

# 4. CONFUSION MATRIX
cm = confusion_matrix(true_classes, predicted_classes)

plt.figure(figsize=(8, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_labels,
    yticklabels=class_labels
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()

# 5. SAVE FINAL MODEL
model.save("cnn_intel_final_model.keras")
print("Final model saved successfully!")

# 6. SAMPLE VISUAL PREDICTIONS

plt.figure(figsize=(12, 8))

for i in range(6):
    img_batch, label_batch = test_generator[i]

    prediction = model.predict(img_batch)

    pred_class = class_labels[np.argmax(prediction[0])]
    true_class = class_labels[np.argmax(label_batch[0])]

    plt.subplot(2, 3, i + 1)
    plt.imshow(img_batch[0])
    plt.title(f"True: {true_class}\nPred: {pred_class}")
    plt.axis("off")

plt.tight_layout()
plt.show()

# 7. FINAL SUMMARY OUTPUT
print("\nFINAL RESULTS SUMMARY")
print("----------------------")
print("Test Accuracy:", test_acc)
print("Model saved as: cnn_intel_final_model.h5")
print("Classes:", class_labels)
```
